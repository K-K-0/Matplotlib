```python
import matplotlib.pyplot as plt
import numpy as np 
```

## 1. Line Plot Basics

Create a simple line plot for x = [1,2,3,4,5] and y = [2,4,6,8,10]. Add title and axis labels.

```python
x = [1,2,3,4,5]
y = [2,4,6,8,10]

fig, ax = plt.subplots()
ax.plot(x,y)
plt.tight_layout()

plt.show()


```

## 2. Multiple Lines

Plot two lines on the same graph: y1 = x^2 and y2 = x^3 for x = range(1, 6). Add legend.

```python
x = range(1,6)
y1 = [i^2 for i in x]
y2 = [i^3 for i in x]


fig, ax = plt.subplots()
ax.plot(y1)
ax.plot(y2)
plt.tight_layout()

plt.show()
```

## 3. Bar Chart

Create a bar chart showing marks of students: Names = ['A', 'B', 'C', 'D'], Marks = [75, 88, 62, 90].

```python
Names = ['A', 'B', 'C', 'D']
Marks = [75, 88, 62, 90]

fig, ax = plt.subplots()
ax.plot(Names, Marks)
plt.tight_layout()
plt.show()
```

## 4. Scatter Plot

Create a scatter plot for random 20 points. Change marker size and style.

```python
x = np.random.random(20)
y = np.random.random(20)

# plt.style.use('_mpl-gallery')
fig, ax = plt.subplots()

ax.scatter(x,y, s=50, c='green')
# plt.tight_layout()
plt.show()
```

## 5. Histogram

Generate 100 random numbers and plot a histogram with 10 bins.

```python
x = np.random.rand(100)
fig, ax = plt.subplots()
ax.hist(x, bins=10, edgecolor='white')
# plt.tight_layout()
plt.show()
```

## 6. Pie Chart

Create a pie chart showing time spent: Study=5, Sleep=8, Play=3, Others=8.

```python
value = [5,8,3,8]
lab = ['Study', 'Sleep', 'Play', 'Others']

plt.style.use('_mpl-gallery-nogrid')

fig, ax = plt.subplots()
ax.pie(value, labels=lab, radius=2, center=(3,3),wedgeprops={"linewidth": 1, "edgecolor": "white"})
# plt.tight_layout()

plt.show()
```

## 7. Subplots

Create a figure with 2 subplots: one line plot and one bar chart.

```python
value = [5,8,3,8,7,4,9]

fig, ax1 = plt.subplots()
fig, ax2 = plt.subplots()

ax1.plot(value)
ax2.bar(value, height=value)


```

## 8. Grid and Style

Plot any graph and enable grid. Try using a Matplotlib style like 'ggplot'.

```python
x = np.random.random(20)
y = np.random.random(20)

plt.style.use('ggplot')
fig, ax = plt.subplots()

ax.scatter(x,y, s=50, c='green')
# plt.tight_layout()
plt.show()
```

## 9. Save Figure

Create any plot and save it as 'my_plot.png'.

```python
x = np.random.random(20)
y = np.random.random(20)

plt.style.use('_mpl-gallery')
fig, ax = plt.subplots()

ax.scatter(x,y, s=50, c='green')
plt.savefig('my_plot.png', dpi=900, bbox_inches='tight')
plt.show()
```
