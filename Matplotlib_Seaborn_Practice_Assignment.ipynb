## 📁 Dataset Setup

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = sns.load_dataset('titanic')
df.head()
```

## 🧩 Part 1: Distribution Plots

### Q1. Histogram (Hist Plot)
- Plot a histogram for a numerical column
- bins = 30, stat='count'

**Write interpretation below**

```python
stat = df['age'].dropna()
fig, ax = plt.subplots()

ax.hist(stat, bins=30, edgecolor='white')
ax.set_ylabel('Count')
ax.set_xlabel('Age')

plt.tight_layout()
plt.show()
```

### Q2. KDE Plot
- Overlay KDE on histogram

**Write interpretation below**

```python
stat = df['age'].dropna()
fig, ax = plt.subplots()

sns.histplot(stat, bins=30, edgecolor='white', kde=True)
ax.set_ylabel('Count')
ax.set_xlabel('Age')

plt.tight_layout()
plt.show()
```

### Q3. ECDF Plot
- Create ECDF plot

**Write interpretation below**

```python
stat = df['fare'].dropna()
fig, ax = plt.subplots()

sns.ecdfplot(stat)
ax.set_ylabel('Count')
ax.set_xlabel('Age')

plt.tight_layout()
plt.show()
```

### Q4. Rug Plot
- Add rug plot

**Write interpretation below**

```python
stat = df['age'].dropna()
fig, ax = plt.subplots()

sns.rugplot(stat)
ax.set_ylabel('Count')
ax.set_xlabel('Age')

plt.tight_layout()
plt.show()
```

### Q5. Wrapper API – displot
- Use sns.displot with different kinds

**Write interpretation below**

```python
age = df['age'].dropna()
# fig, ax = plt.subplots()

sns.displot(age, bins=30, kind='hist')


plt.tight_layout()
plt.show()
```

## 🔗 Part 2: Relational Plots

### Q6. Scatter Plot
- Use hue

**Write interpretation below**

```python
age = df['age'].dropna()
# fig, ax = plt.subplots()

sns.scatterplot(data=df, x='age', y='fare', hue='sex' )

plt.tight_layout()
plt.show()
```

### Q7. Line Plot
- Show trend using mean aggregation

**Write interpretation below**

```python
age = df.groupby('class')['fare'].mean().reset_index()
# fig, ax = plt.subplots()

sns.lineplot(data=age, x='class', y='fare')


# plt.tight_layout()
plt.show()
```

### Q8. Wrapper API – relplot
